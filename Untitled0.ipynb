```python
import numpy as np
import urllib.request
import os
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, accuracy_score
import gradio as gr

print("TensorFlow:", tf.__version__)
```

```python
CATEGORIES = ["cat", "apple", "tree", "star", "house"]
DATA_DIR = "quickdraw_data"
os.makedirs(DATA_DIR, exist_ok=True)

BASE_URL = "https://storage.googleapis.com/quickdraw_dataset/full/numpy_bitmap/{}.npy"

for cat in CATEGORIES:
    path = os.path.join(DATA_DIR, f"{cat}.npy")
    if not os.path.exists(path):
        print(f"Descargando {cat}...")
        urllib.request.urlretrieve(BASE_URL.format(cat), path)
    else:
        print(f"{cat} ya descargado.")

print("Descarga completa.")
```

```python
SAMPLES_PER_CLASS = 4000  # suficiente para superar 75% sin tardar demasiado en entrenar

X_list = []
y_list = []

for idx, cat in enumerate(CATEGORIES):
    arr = np.load(os.path.join(DATA_DIR, f"{cat}.npy"))
    np.random.shuffle(arr)
    arr = arr[:SAMPLES_PER_CLASS]
    X_list.append(arr)
    y_list.append(np.full(len(arr), idx))
    print(f"{cat}: {len(arr)} imagenes")

X = np.concatenate(X_list, axis=0).astype("float32") / 255.0
y = np.concatenate(y_list, axis=0)

X = X.reshape(-1, 28, 28, 1)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.15, random_state=42, stratify=y
)

print("Train:", X_train.shape, "Test:", X_test.shape)
```

```python
fig, axes = plt.subplots(1, 5, figsize=(12, 3))
for i, ax in enumerate(axes):
    idx = np.random.randint(0, len(X_train))
    ax.imshow(X_train[idx].squeeze(), cmap="gray")
    ax.set_title(CATEGORIES[y_train[idx]])
    ax.axis("off")
plt.show()
```

```python
model = models.Sequential([
    layers.Input(shape=(28, 28, 1)),
    layers.Conv2D(32, (3, 3), activation="relu", padding="same"),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation="relu", padding="same"),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation="relu", padding="same"),
    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dropout(0.3),
    layers.Dense(len(CATEGORIES), activation="softmax"),
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python
history = model.fit(
    X_train, y_train,
    validation_split=0.15,
    epochs=12,
    batch_size=128,
    verbose=1
)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(history.history["accuracy"], label="train")
axes[0].plot(history.history["val_accuracy"], label="val")
axes[0].set_title("Accuracy")
axes[0].legend()

axes[1].plot(history.history["loss"], label="train")
axes[1].plot(history.history["val_loss"], label="val")
axes[1].set_title("Loss")
axes[1].legend()
plt.show()
```

```python
test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)
print(f"Accuracy en test: {test_acc*100:.2f}%")

y_pred = np.argmax(model.predict(X_test, verbose=0), axis=1)

cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=CATEGORIES)
fig, ax = plt.subplots(figsize=(6, 6))
disp.plot(ax=ax, cmap="Blues", colorbar=False)
plt.title(f"Matriz de confusion (accuracy: {test_acc*100:.2f}%)")
plt.savefig("matriz_confusion.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
model.save("modelo_adivina_mi_dibujo.h5")
print("Modelo guardado como modelo_adivina_mi_dibujo.h5")
```

```python
def predecir(imagen):
    if imagen is None:
        return {cat: 0.0 for cat in CATEGORIES}

    if isinstance(imagen, dict):
        img = imagen.get("composite", imagen.get("image"))
    else:
        img = imagen

    if img is None:
        return {cat: 0.0 for cat in CATEGORIES}

    img = np.array(img)
    if img.ndim == 3:
        img = img.mean(axis=2)

    from PIL import Image
    img_pil = Image.fromarray(img.astype("uint8")).resize((28, 28))
    img_arr = np.array(img_pil).astype("float32") / 255.0

    if img_arr.mean() > 0.5:
        img_arr = 1.0 - img_arr

    img_arr = img_arr.reshape(1, 28, 28, 1)
    pred = model.predict(img_arr, verbose=0)[0]

    top3 = np.argsort(pred)[::-1][:3]
    return {CATEGORIES[i]: float(pred[i]) for i in top3}


demo = gr.Interface(
    fn=predecir,
    inputs=gr.Sketchpad(type="numpy", label="Dibuja aqui"),
    outputs=gr.Label(num_top_classes=3, label="Predicciones (top 3)"),
    live=True,
    title="Adivina mi Dibujo",
    description=f"Dibuja uno de: {', '.join(CATEGORIES)}. La prediccion se actualiza mientras dibujas."
)

demo.launch(share=True, debug=True)
```
